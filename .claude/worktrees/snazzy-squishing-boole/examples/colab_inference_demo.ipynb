```python
# @title Inference Parameters { display-mode: "form" }
# Note: Colab form fields are single-line. For sequences >200 residues,
# paste the full one-letter amino acid string — it will scroll horizontally.
sequence = 'MKTAYIAKQRQISFVKSHFSRQLEERLGLIEVQAPILSRVGDGTQDNLSGAEKAVQVKVKALPDAQFEVVHSLAKWKRQTLGQHDFSAGEGLYTHMKALRPDEDRLSPLHSVYVDQWDWERVMGDGERQFSTLKSTVEAIWAGIKATEAAVSEEFGLAPFLPDQIHFVHSQELLSRYPDLDAKGRERAIAKDLGAVFLVGIGGKLSDGHRHDVRAPDYDDWSTPSELGHAGLNGDILVWNPVLEDAFELSSMGIRVDADTLKHQLALTGDEDRLELEWHQALLRGEMPQTIGGGIGQSRLTMLLLQLPHIGQVQAGVWPAAVRESVPSLL'  # @param {type:"string"}
mode = "score"  # @param ["score", "sample"]
temperature = 0.5  # @param {type:"slider", min:0.1, max:1.0, step:0.05}
num_samples = 1  # @param {type:"integer"}
```

## Run Inference

Click **Run Inference** to execute with the parameters set above.

```python
import ipywidgets as widgets
from IPython.display import display

run_button = widgets.Button(
    description='Run Inference',
    button_style='info',
    tooltip='Click to run inference with parameters above'
)
output_area = widgets.Output()

def run_inference(button):
    """Execute inference using the #@param variables defined above."""
    output_area.clear_output(wait=True)
    with output_area:
        try:
            seq = sequence.strip()
            if not seq:
                print("Error: Please enter a protein sequence.")
                return
            if len(seq) < 10:
                print(f"Warning: Sequence is very short ({len(seq)} residues).")

            print(f"\n=== PrxteinMPNN {mode.upper()} ===")
            print(f"Sequence length: {len(seq)} residues")
            print(f"Temperature: {temperature}")

            if mode == 'score':
                print("\nScoring sequence...")
                spec = ScoringSpecification(
                    sequence=seq,
                    temperature=temperature,
                )
                result = score(spec)
                print("\nScore result:")
                print(f"  Input sequence: {seq}")
                if hasattr(result, 'scores'):
                    print(f"  Scores: {result.scores}")
                else:
                    print(f"  Result: {result}")

            elif mode == 'sample':
                print(f"\nSampling {num_samples} sequence(s)...")
                spec = SamplingSpecification(
                    sequence=seq,
                    temperature=temperature,
                    num_samples=num_samples,
                )
                results = sample(spec)
                print("\nSample results:")
                print(f"  Input sequence: {seq}")
                if isinstance(results, (list, tuple)):
                    for i, s in enumerate(results):
                        print(f"  Sample {i+1}: {s}")
                else:
                    print(f"  Generated: {results}")

            print("\nInference complete!")

        except Exception as e:
            print(f"Error during inference: {type(e).__name__}")
            print(f"Message: {str(e)}")
            print("\nTip: Ensure sequence is valid and model weights are accessible.")

run_button.on_click(run_inference)
display(run_button)
display(output_area)
```

## About PrxteinMPNN

PrxteinMPNN is a functional interface for ProteinMPNN, a deep learning model for protein design.

**Two main modes:**
- **Scoring**: Evaluate the likelihood of a protein sequence
- **Sampling**: Generate new protein sequences conditioned on a context sequence

**Key parameters:**
- **Temperature**: Controls diversity. Lower values (0.1) = more deterministic; higher values (1.0) = more random
- **Sequence**: The input protein sequence (one-letter amino acid codes)
